```python
# Load env variable
from dotenv import load_dotenv

load_dotenv()
```

```python
# Create an API client
from anthropic import Anthropic

client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
# Add helper functions to append conversation
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)


def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)



def chat(messages):
    message = client.messages.create(
    model=model,
    max_tokens=1000,
    messages=messages
    )

    return message.content[0].text
```

```python
# Start with an empty messages list
messages = []

# Use a while True loop to keep asking the user for questions
while True:
    # Ask user for question
    user_input = input("What is your question, my dear Madamoiselle? Press q to quit.")
    
    # Condition to break the while loop
    if user_input == 'q':
        print("\n The user has quit the chatbot!")
        break

    # Print the user input
    print("--------")
    print(">", user_input)

    # Add user message to messages list
    add_user_message(messages, user_input)

    # Get response from Claude for user message
    answer = chat(messages)
    
    # Print Claude's response 
    print(f"Madamoiselle, here's the answer: {answer}")

    # Add response to messages list
    add_assistant_message(messages, answer)


```
